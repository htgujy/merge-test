코드 2-27

```python
a = int(input('정수: '))
b = int(input('정수: '))
print(f'{a} + {b} = {a+b}')
print(f'{a} - {b} = {a-b}')
print(f'{a} * {b} = {a*b}')
print(f'{a} / {b} = {a/b}')
print(f'{a} // {b} = {a//b}')
print(f'{a} % {b} = {a % b}')
print(f'{a} ** {b} = {a**b}')
```
