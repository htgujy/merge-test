- **[과제 1] 원의 반지름을 입력받아 원의 넓이를 구하는 프로그램을 작성하시오**

```python
pi=3.14159265358979323846264338
r = int(input('원의 반지름'))
area = pi * r * r
print(f'원의 넓이:{area:.3f}')
```

- **[과제 3] 사용자로부터 입력받은 센티미터 단위의 길이를 미터와 센티미터 단위로 변환하는 프로그램을 작성하시오**
    - (예) 입력값 : 234  /  출력값 : 2미터 34센티미터

```python
cm = int(input('센티미터 단위의 길이:'))
m = cm//100
cm = cm%100
print(f'미터 단위로 환산: {m}미터 {cm}센티미터')
```

- **[과제 4] 사용자로부터 분단위 시간을 입력받아 시간과 분 단위로 변환하는 프로그램을 작성하시오**
    - (예) 입력값 : 145  /  출력값 : 2시간 25분 

```python
m = int(input('분 단위 시간 입력'))
h = m//60
m = m%60
print(f'{h}시간{m}분')
```

- **[과제 5] 사용자로부터 각 음료의 주문 수를 입력받아 주문한 음료의 총 가격을 계산하는 프로그램을 작성하시오**
    - (음료 가격예) 아메리카노 : 2000 / 라떼 : 3000  /  녹차 : 2500

```python
a = int(input('아메리카노:'))
l = int(input('라떼:'))
g = int(input('녹차:'))
total = a*2000 + l*3000 + g*2500
print(f'주문한 음료의 총 가격:{total}원')
```
