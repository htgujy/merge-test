**1. 컴퓨터가 선택한 수 맞히기**

```python
import random

print('1부터 100까지의 수 하나를 정했습니다.')
print('이 숫자를 맞혀보세요.')

answer = random.randint(1, 100)
times = 0
```

```python
while True:
  user = int(input('1~100 숫자 입력: '))
  times += 1
  if user == answer:
    break
  elif user < answer:
    print(f'{user}보다 큽니다.')
  else:
    print(f'{user}보다 작습니다.')
```

```python
print(f'정답입니다! {times}회 만에 맞혔습니다.')
```

**2. 사용자가 생각한 수 맞히기**

```python
print('1~100 중 숫자 하나를 정하세요.')
```

```python
low = 1
high = 100
times = 0
```

```python
while True:
  num = (low+high)//2
  answer = input(f'{num}인가요? OK/UP/DOWN: ')
  times += 1
  if answer == 'OK':
    break
  elif answer == 'UP':
    low = num+1
  elif answer == 'DOWN':
    high = num-1
  else:
    print('잘못 입력했습니다.')
```

```python
print(f'{times}회 만에 맞혔습니다.')
```
