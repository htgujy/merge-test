- **코드 4-13 : while 문의 활용**

```python
a = 1
while a<=3:
  print(f'{a}회 반복')
  a += 1
```

- **코드 4-14 : while 문으로 1부터 5까지 정수의 합 구하기**

```python
sum = 0
a = 1
while a<=5:
  sum += a
  a += 1
print(sum)
```

- **코드 4-15 : 반복문을 빠져나올때 break 사용**

```python
while True:
  answer = input('이름 or 나이 or 종료: ')
  if answer == '종료':
    print('종료하겠습니다.')
    break
  elif answer.isdigit():
    print(f'{answer}세이군요')
  else:
    print(f'{answer}님! 반갑습니다.')
```

- **코드 4-17 : 정수 각 자릿수의 합 구하기**

```python
num = int(input('정수: '))
sum = 0
while num>0:
  sum += num%10
  num = num//10
print(sum)
```
