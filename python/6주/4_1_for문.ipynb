- **응용**

```python
sum = 0
x = int(input('시작 정수입력 : '))
y = int(input('종료 정수입력 : '))

for a in range(x, y+1):
  sum += a
print(sum)
```

- **코드 4-9 : 약수를 구하는 프로그램**

```python
num = int(input('정수: '))
for a in range(1, num+1):
  if num%a == 0:
    print(a, end=' ')
```

- **코드 4-10 : 전 직원의 월급 총액 구하기**

```python
employee_num = 3
total = 0

for i in range(employee_num):
  rank = int(input('직급: '))
  basic_salary = int(input('본봉: '))
  if rank>=2:
    bonus = basic_salary*0.3
  else:
    bonus = basic_salary*0.2
  salary = basic_salary + bonus
  total += salary
print('월급총액:', total)
```
