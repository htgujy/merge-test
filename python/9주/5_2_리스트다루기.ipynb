- **코드 5-5** : 리스트를 항목에 한 개씩 접근하면서 출력하는 방법(1)
  - 인덱스와 반복 구조 이용하기

```python
menus = ['자장면', '짬뽕', '볶음밥']
for i in range(3):
  print(menus[i])
```

- **코드 5-6** : 리스트의 길이를 알려주는 **len()** 함수를 사용하는 방법
  - 항목 수가 달라져도 코드가 동작가능

```python
menus = ['자장면', '짬뽕', '볶음밥']
for i in range(len(menus)):
  print(menus[i])
```

- **코드 5-7** : 리스트를 항목에 한 개씩 접근하면서 출력하는 방법(2)
  - 인덱스 없이 for 문과 리스트 이용하기

```python
menus = ['자장면', '짬뽕', '볶음밥']
for a in menus:
  print(a)
```

- **코드 5-8** : 리스트를 항목에 한 개씩 접근하면서 출력하는 방법(3)
  - **enumerate()** 함수 : 인덱스와 항목을 모두 반환

```python
menus = ['자장면', '짬뽕', '볶음밥']
for i, a in enumerate(menus):
  print(i, a)
```

- **리스트 슬라이싱** :letters[1:3] ==> ['B', 'C' ]
> 

```python
letters=['A', 'B', 'C', 'D', 'E']
print(letters[1:3])
print(letters[0:4])
print(letters[:4])
print(letters[2:])
print(letters[1:-1])
```

- **코드 5-9** : 항목 추가하기

```python
pl = []
pl.append('파이썬')
pl.append('C')
print(pl)
```

- **코드 5-10** : 항목 추가하기
  - **insert(위치, 값) 함수** : 지정한 리스트이 위치에 값을 추가

```python
pl = ['파이썬', '자바']
pl.insert(1, 'C')
print(pl)
```

- **코드 5-11** : 항목 추가하기
  - **리스트1.extend(리스트2) 함수** : 리스트1 뒤에 리스트2의 항목을 추가코드 5-11

```python
pl = ['파이썬', 'C']
pl.extend(['C++', '자바'])
print(pl)
```

- **코드 5-12** : 항목 삭제하기
  - **리스트.pop(인덱스) 함수** : 리스트의 인덱스 위치에 있는 항목을 삭제. 인덱스를 생략하면 마지막 항목을 삭제

```python
pets = ['dog', 'cat', 'hamster', 'hedgehog', 'parrot', 'lizard']
select1 = pets.pop(3)
print(select1)
select2 = pets.pop()
print(select2)
print(pets)
```

- **코드 5-13** : 항목 삭제하기
  - **del 키워드** : 삭제한 항목을 반환하지 않음

```python
pets = ['dog', 'cat', 'hamster', 'hedgehog', 'parrot', 'lizard']
del pets[3], pets[-1]
print(pets)
```

- **코드 5-14** :  항목 삭제하기
  - **리스트.remove(값)** 함수 : 리스트에서 첫번째 위치의 '값'을 삭제

```python
pets = ['dog', 'cat', 'hamster', 'hedgehog', 'parrot', 'lizard']
pets.remove('hedgehog')
print(pets)
```

- **코드 5-15** : 항목 삭제하기
  - **clear()** 함수 : 리스트의 모든 항목을 삭제

```python
pets = ['dog', 'cat', 'hamster', 'hedgehog', 'parrot', 'lizard']
pets.clear()
print(pets)
```

- **코드 5-16** : 항목 재배열하기
  - **reverse()** 함수 : 리스트 항목을 역순으로 뒤집는다.

```python
nations = ['한국', '미국', '영국', '일본']
nations.reverse()
print(nations)
```

- **코드 5-17** : 항목 재배열하기
  - **sort()** 함수 : 리스트 항목을 정렬한다. **reverse=True**로 설정하면 내림차순

```python
nations = ['한국', '미국', '영국', '일본']
nations.sort()
print(nations)
nations.sort(reverse=True)
print(nations)
```

- **코드 5-18** : 리스트 탐색하기
  - **in** 연산자 : 리스트에 특정 값이 있는지를 알려줌 <값 in 리스트>
  - **index()** 함수 : 해당 항목의 인덱스를 알려줌  <리스트.index(값)>
> 

```python
# 영어로 된 월 이름이 숫자로 몇 월인지 알려주는 프로그램

months = ['January','February','March','April','May','June',
          'July','August','September','October','November','December']
answer = input('월(영어): ')

if answer in months:
  print(months.index(answer)+1)
else:
  print('잘못 입력했습니다.')
```

- **코드 5-19** : 랜덤 선택
  - **random.choice(리스트)** : 리스트에서 항목 한개를 무작위로 선택
  - **random.choice(리스트, k=갯수)** : 리스트에서 항목을 설정한 갯수만큼 **중복을 허용**해 무작위로 선택
  - **random.sample(리스트, 갯수)** : 설정한 갯수만큼 리스트에서 항목을 **중복 없이** 무작위로 선택

```python
import random

colors = ['ivory', 'lavender', 'lightcyan', 'lightyellow', 'palegreem']
print(random.choice(colors))
print(random.sample(colors,3))
print(random.choices(colors,k=3))
```

- **코드 5-20** : 랜덤 재배열
  - **random.shuffle(리스트)** : 리스트 항목들을 무작위로 재배열(섞는다)

```python
# import random

colors = ['ivory', 'lavender', 'lightcyan', 'lightyellow', 'palegreem']
random.shuffle(colors)
print(colors[:3])
```

- **코드 5-21** : 통계 관련 내장 함수
  - **max(리스트), min(리스트), sum(리스트)** 함수 : 최대값, 최소값, 항목들의 합을 구한다.

```python
scores = [85, 95, 70, 75, 99, 93, 88]

print('최댓값:', max(scores))
print('최솟값:', min(scores))
print('합:', sum(scores))
print('평균:', sum(scores)/len(scores))
```
