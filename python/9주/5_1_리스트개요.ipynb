## 5.1 리스트 개요
> 

```python
prices = [1500, 3200, 2500, 4000, 5500]
print(prices)
```

```python
print(prices[0])
print(prices[1])
print(prices[2])
print(prices[3])
print(prices[4])
```

- 각 상품(prices)의 가격을 5% 인상하고 인상된 가격을 출력하는 코드
  - **len()** : 객체의 길이(요소 개수)를 정수(int)로 반환하는 함수 

```python
for i in range(len(prices)):
    prices[i] = prices[i] + prices[i]*0.05

print(prices)
```

- **코드 5-1** : 반복문으로 리스트 생성하기

```python
# 1부터 10까지의 정수를 세제곱한 수를 cubes 리스트에 저장하는 프로그램

cubes = []
for x in range(1, 11):
  cubes.append(x**3)
print(cubes)
```

- 리스트 내포(list comprehension)
  - 형식 : **[식 for 변수 in 반복가능한 것 if 조건식]**
  - 대괄호안에 for문과 조건문을 내포시켜 리스트를 한줄로 생성하는 구문
  - if 조건식은 생략 가능

```python
# 코드 5-1의 리스트 내포 예

cubes = [x**3 for x in range(1,11)]
print(cubes)
```

- **코드 5-2** : 리스트 예

```python
# 1부터 100까지 수 중 7의 배수를 sequence 리스트에 저장하는 프로그램

sequence = []
for x in range(1, 101):
  if x%7 == 0:
    sequence.append(x)
print(sequence)
```

> 

```python
# 코드 5-2의 리스트 내포 예

sequence = [x for x in range(1, 101) if x%7 ==0 ]
print(sequence)
```

- **코드 5-3**  리스트 예

```python
score = [71, 80, 60, 90, 65]
result = []

for a in score:
  if a >= 70:
    result.append('합격')
  else:
    result.append('불합격')
print(result)
```

- **코드 5-4** : 코드 5-3의 리스트 내포 예
> 

```python
score = [71, 80, 60, 90, 65]
result = ['합격' if a>=70 else '불합격' for a in score]
print(result)
```
