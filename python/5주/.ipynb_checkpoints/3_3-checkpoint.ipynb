코드 3-11

```python
delivery_fee = 2000
level = input('회원 등급(우수 또는 일반): ')
price = int(input('구매액: '))

if level=='우수':
  print('결제금액:', price)
elif price>=20000:
  print('결제금액:', price)
else:
  print('결제금액:', price+delivery_fee)
```

코드 3-12

```python
delivery_fee = 2000
level = input('회원 등급(우수 또는 일반): ')
price = int(input('구매액: '))

if level=='우수' or price>=20000:
  print('결제금액:', price)
else:
  print('결제금액:', price+delivery_fee)
```

코드 3-13

```python
a = int(input('정수: '))
if a%3==0 and a%5==0:
  print('3과 5의 배수')
else:
  print('3과 5의 배수 아님')
```
