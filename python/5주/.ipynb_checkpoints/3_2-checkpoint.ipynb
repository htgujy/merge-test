```python
import random
a = random.randint(1,100)
b = random.randint(1,100)
if a < b:
  a, b = b, a
print(a, b)
```

코드 3-5

```python
height1 = int(input('키: '))
height2 = int(input('키: '))
if height1 > height2:
  print('키 차이:', height1-height2)
else:
  print('키 차이:', height2-height1)
```

```python
a = int(input('정수를 입력하세요:'))
if a > 0:
    print('양수입니다')
elif a== 0:
    print('0입니다.')
else: 
    print('음수입니다.')
```

코드 3-6

```python
fee = 5000
age = int(input('나이: '))
if age < 8:
  print('입장료:', fee*0.7)
elif age < 60:
  print('입장료:', fee)
else:
  print('입장료:', fee*0.8)
```

```python
n = int(input('정수를 입력하세요:'))
if n%2 == 0:
    print(f'{n}: 짝수')
else:
    print(f'{n}: 홀수')
```
